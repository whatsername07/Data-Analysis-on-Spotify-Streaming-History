# Spotify History Data Cleaning
## Loading into dataframe

```python
import pandas as pd

fullHistory = pd.read_csv("./Spotify Extended Streaming History/raw/full-history.csv")
fullHistory.sample(10)
```

```python
# drop all podcast and audiobook rows
fullHistory = fullHistory[~fullHistory["episode_name"].notna()]
fullHistory = fullHistory[~fullHistory["audiobook_uri"].notna()]

# drop podcasts and audiobooks columns
fullHistory = fullHistory.drop(
    columns=[
        "episode_name",
        "episode_show_name",
        "spotify_episode_uri",
        "audiobook_title",
        "audiobook_chapter_uri",
        "audiobook_uri",
        "audiobook_chapter_title",
    ]
)
```

```python
fullHistory.sample(10)
```

```python
# drop irrelevant columns
fullHistory = fullHistory.drop(
    columns=[
        "platform",
        "spotify_track_uri",
        "shuffle",
        "offline",
        "offline_timestamp",
        "incognito_mode"
    ]
)
```

```python
fullHistory['ts'] = pd.to_datetime(fullHistory['ts'])
```

```python
fullHistory.to_csv("./Spotify Extended Streaming History/cleaned/cleaned-history.csv", index=False)
```

```python
# top artist by play count
topArtists = fullHistory["master_metadata_album_artist_name"].value_counts().head(10)
print("Top 10 Artists by Play Count:")
print(topArtists)
```

## Function to Split History by Year

```python
def historySplit(df, year):
    return df[df['ts'].dt.year == year]
```

```python
history2023 = historySplit(fullHistory, 2023)
```
