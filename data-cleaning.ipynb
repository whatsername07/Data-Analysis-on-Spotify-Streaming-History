# Spotify History Data Cleaning
## Loading into dataframe

```python
import pandas as pd

fullHistory = pd.read_csv("./Spotify Extended Streaming History/full-history.csv")
fullHistory.sample(10)
```

```python
# drop all podcast and audiobook rows
fullHistory = fullHistory[~fullHistory["episode_name"].notna()]
fullHistory = fullHistory[~fullHistory["audiobook_uri"].notna()]

# drop podcasts and audiobooks columns
fullHistory = fullHistory.drop(
    columns=[
        "episode_name",
        "episode_show_name",
        "spotify_episode_uri",
        "audiobook_title",
        "audiobook_chapter_uri",
        "audiobook_uri",
        "audiobook_chapter_title",
    ]
)
```
